# 01 — HH-RLHF Data Exploration

Initial exploration of the Anthropic HH-RLHF dataset: structure, size, and chosen/rejected pair format.

```python
from datasets import load_dataset
```

## Load the dataset

```python
hh = load_dataset('Anthropic/hh-rlhf')
print(hh)
```

## Dataset size

```python
for split, ds in hh.items():
    print(f"{split}: {len(ds):,} examples")
```

## First example (raw)

```python
example = hh['train'][0]
print(example)
```

## Chosen / Rejected pair — readable view

```python
def display_pair(example, index=0):
    print(f"{'='*60}")
    print(f"Example {index}")
    print(f"{'='*60}")
    print("\n--- CHOSEN ---")
    print(example['chosen'])
    print("\n--- REJECTED ---")
    print(example['rejected'])
    print(f"{'='*60}\n")

display_pair(hh['train'][0], index=0)
```

## Column schema

```python
print(hh['train'].features)
```

## Basic length stats on chosen vs rejected responses

```python
import pandas as pd

sample = hh['train'].select(range(1000))
df = pd.DataFrame({
    'chosen_len': [len(x['chosen']) for x in sample],
    'rejected_len': [len(x['rejected']) for x in sample],
})
df['chosen_longer'] = df['chosen_len'] > df['rejected_len']

print(df.describe())
print(f"\nChosen response longer than rejected: {df['chosen_longer'].mean():.1%} of the time")
```
